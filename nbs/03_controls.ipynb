# Controls and feedback

> Attribute-based CSS skins for buttons, form inputs, field groups, cards, badges, avatars, status indicators, and data tables.

AnswerUI applies component styles through HTML attributes such as `[btn]`, `[input]`, and `[card]`, keeping native focus, keyboard activation, form submission, and validation intact. Add `[btn]` to an `<a>` element to style a link like a button without changing its navigation semantics. Omit `[btn]` on ordinary links to keep default link styling.

## Buttons

Add `[btn]` to a `<button>` or `<a>` for a primary accent button with `--bevel` and `--raised` shadows. Set `btn-variant="secondary"` for a bordered `--color-surface` button, or `btn-variant="quiet"` for a transparent button that highlights on hover. Set `btn-variant="editable"` for a full-width, dashed-border trigger with a text cursor for click-to-edit values. Change button size with `btn-size="sm"` or `btn-size="lg"`, and add `[btn-icon]` for square icon-only buttons.

Setting `aria-pressed="true"` or `aria-expanded="true"` switches a button to `--ui-held-face` and `--ui-held-shadow` (`--bevel-in` and `--sunken` on primary and secondary buttons). Adding `disabled` lowers opacity to 50% and removes the shadow. The CSS never binds click handlers, toggles ARIA attributes, or turns a button into a link. Wrapper components default to `type="button"`, whereas plain `<button btn>` elements follow native form defaults. On `[btn][copy]` and `[btn][copyfor]`, setting `[copied]` swaps the first `<svg><use>` icon for the second.

Setting `aria-busy="true"` hides child `span` and `svg` elements and shows a centred spinner immediately. With HTMX's `.htmx-request` class, the same busy state appears after `--busy-delay` (default `200ms`). Fast requests finish before the spinner appears. Under `prefers-reduced-motion: reduce`, button transitions are disabled and the spinner slows to `2s`.

```
#| hide
#| asset styles.css
_css = r'''
@layer components {
@scope (:root, [ui]) to ([unstyle]) {
  :where(:scope, *)[btn] {
    --face: var(--color-accent);
    --ui-held-face: var(--v2);
    --ui-held-shadow: var(--bevel-in), var(--sunken);
    --ui-pad-block: var(--space-2);
    --ui-pad-inline: var(--space);
    --ui-press: 0px;
    --busy-delay: 200ms;
    position: relative;
    display: inline-flex; align-items: center; justify-content: center; gap: var(--space-2);
    padding: calc(var(--ui-pad-block) + var(--ui-press)) calc(var(--ui-pad-inline) - var(--ui-press))
      calc(var(--ui-pad-block) - var(--ui-press)) calc(var(--ui-pad-inline) + var(--ui-press));
    border: 1px solid transparent; border-radius: var(--radius);
    background: var(--face); color: var(--color-on-accent); box-shadow: var(--bevel), var(--raised);
    font: inherit; font-weight: var(--weight1); line-height: var(--leading-1);
    text-align: center; text-decoration: none; cursor: pointer;
    transition: background var(--motion) var(--ease), box-shadow var(--motion) var(--ease), padding calc(var(--motion) / 2) var(--ease), opacity var(--motion) var(--ease);

    & > :where(span) { display: inline-flex; align-items: center; justify-content: inherit; gap: var(--space-2); }
    &:where(:hover:not(:disabled)) { --face: var(--v1); }
    &:where(:active:not(:disabled)) { --ui-press: calc(1px * var(--relief)); }
    &:where([btn-size=sm]) { --ui-pad-block: var(--space-3); --ui-pad-inline: var(--space-2); font-size: var(--text-1); }
    &:where([btn-size=lg]) { --ui-pad-block: var(--space-1); --ui-pad-inline: var(--space1); font-size: var(--text1); }
    &:where([btn-variant=secondary]) {
      --face: var(--color-surface); --ui-held-face: var(--color-surface); color: var(--color-text); border-color: var(--color-border);
      &:where(:hover:not(:disabled)) { --face: var(--v-3); }
      &:where([aria-pressed=true], [aria-expanded=true], [aria-busy=true]) { --ui-held-face: var(--v-3); color: var(--v2); }
      &:where(.htmx-request) { --ui-held-face: var(--v-3); &::after { color: var(--v2); } }
    }
    &:where([btn-variant=quiet], [btn-variant=editable]) {
      --face: transparent; --ui-held-face: color-mix(in oklch, currentColor 12%, transparent); --ui-held-shadow: none;
      --ui-press: 0px; color: inherit; box-shadow: none;
      &:where(:hover:not(:disabled)) { --face: var(--tint); }
    }
    &:where([btn-variant=editable]) {
      --ui-pad-inline: var(--space-1); inline-size: 100%; justify-content: start; border-color: var(--n); border-style: dashed;
      font-weight: inherit; text-align: start; cursor: text;
    }
    &:where(:is(:active, [aria-busy=true], [aria-pressed=true], [aria-expanded=true]):not(:disabled)) {
      --face: var(--ui-held-face); box-shadow: var(--ui-held-shadow);
    }
    &:where(:disabled) { opacity: .5; box-shadow: none; cursor: default; }
    &:where([aria-busy=true]) { -webkit-text-fill-color: transparent; }
    &:where([aria-busy=true], .htmx-request) {
      & > :where(span, svg) { opacity: 0; }
      &::after {
        content: ''; position: absolute; inset: 0; margin: auto; inline-size: 1.25em; block-size: 1.25em;
        border: 2px solid currentColor; border-inline-end-color: transparent; border-radius: var(--radius-full);
        animation: ui-spin calc(var(--motion) * 5) linear infinite;
      }
    }
    &:where(.htmx-request:not([aria-busy=true])) {
      animation: ui-busy-held 0s linear var(--busy-delay) both;
      & > :where(span, svg) { transition: opacity var(--motion) var(--ease) var(--busy-delay); }
      &::after { animation: ui-spin calc(var(--motion) * 5) linear infinite, ui-busy-in var(--motion) var(--ease) var(--busy-delay) both; }
    }
    &:where([btn-icon]) {
      --ui-pad-inline: var(--ui-pad-block);
      & > :where(span) { inline-size: 1.25em; block-size: 1.25em; }
    }
    &:is([copyfor], [copy]) {
      & > svg > use:last-child { display: none; }
      &[copied] > svg > use:first-child { display: none; }
      &[copied] > svg > use:last-child { display: inline; }
    }
    @media (prefers-reduced-motion: reduce) {
      transition: none;
      &:where([aria-busy=true], .htmx-request)::after { animation-duration: 2s, 0s; }
    }
  }
  @keyframes ui-busy-held { to { --face: var(--ui-held-face); box-shadow: var(--ui-held-shadow); -webkit-text-fill-color: transparent; } }
  @keyframes ui-busy-in { from { opacity: 0; } }
}
}
'''
```

## Form controls, fields, and groups

Add `[input]`, `[textarea]`, or `[select]` for full-width sunken fields (`--bevel-in` and `--sunken` on `--color-background`); `[input]` also works on a container such as `<div input>` for mounted third-party inputs (for example, Stripe Elements). Browser spin buttons are hidden on `[input][type="number"]`. `[textarea]` starts at `5rem` high, resizes vertically, and grows with its content via `field-sizing: content`. When a text, select, or choice control matches `:user-invalid` or `[aria-invalid="true"]`, its border uses `--hue-danger`. Connect labels and hints with standard HTML `for`, `id`, and `aria-describedby` attributes.

Add `[checkbox]`, `[radio]`, `[switch]`, or `[range]` to style native inputs while keeping native control behaviour. `[checkbox]` supports both `:checked` and `:indeterminate` states. Set `--ui-range-fill` (default `0%`) on `[range]` to colour the filled track. Both `[switch]` and `[range]` reverse direction in right-to-left layouts. Under `prefers-reduced-motion: reduce`, control transitions are disabled. Under `forced-colors: active`, `[checkbox]`, `[radio]`, `[switch]`, and `[range]` revert to `appearance: auto`.

Wrap a `<label>`, a `[field-control]` container, and an optional `<small>` hint in `[field]`. Adding any non-neutral `[ui-tone]` to `<small>` or `<p>` colours the text with `--v2` both inside and outside `[field]`. Add `[field-inline]` to place a checkbox, radio, or switch beside its label with the hint underneath. A `<label>` inside `[field]` is dimmed when any descendant is `:disabled`. Use `[fieldset]` for a bordered `<fieldset>` and bold `<legend>`. Wrap adjacent `[btn]`, `[input]`, `[textarea]`, or `[select]` elements in `[group]` to join them with shared borders and outer corner rounding. Focusing an input inside `[group]` draws one `--ring` outline around the group container. Add `group-variant="inset"` to place a compact `[btn]` inside a single bordered input box (use `[cluster]` for separate controls).

## Surfaces, status, and tables

Add `[card]` for a raised `--color-surface` container with `--space` padding and a border. Combine `[card]` with `[stack]` or another layout attribute when you need spacing between its children. Add `[badge]` for an inline status label tinted by the active `[ui-tone]`, or `badge-variant="secondary"` for a `--color-surface` badge with an inset border. Add `[avatar]` for a circular image or initials box sized by `--avatar-size` (default `2.5rem`). When `[avatar]` is the only child of `[btn]`, the button becomes a circular `1px`-padded `--color-surface` frame.

Use `[alert]` with `[ui-tone]` for persistent notices instead of transient button states. Inside `[alert]`, a direct child `[stack]` fills the remaining row width with `--space-2` gaps. Add `[progress]` to style a native `<progress>` bar at full width with `--color-accent`. Add `[spinner]` to show a `1em` spinning indicator before its label. Under `prefers-reduced-motion: reduce`, the animation stops. Add `[skeleton]` for a full-width placeholder bar (`1em` high) in `--n-1`.

Add `[table]` to a `<table>` (`table[table]`) for row borders, start-aligned cells, bold headers and footer headings, and a bottom caption. Add `[num]` to numeric `<th>` and `<td>` cells for end alignment and tabular digits (`tabular-nums`). Wrap `table[table]` in `[table-frame]` for a bordered, horizontally scrolling container with `--color-surface` headers and `white-space: nowrap` cells. Give `[table-frame]` `role="region"`, `tabindex="0"`, and `aria-label` so keyboard users can scroll it. Plain `<table>` elements inside `[prose]` use the prose table styles instead.
