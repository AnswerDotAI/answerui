## Responsive navbars, toolbars, and breadcrumbs

Translated labels and dynamic items can overflow a navbar at widths where a fixed breakpoint does not trigger. `watchInlineFit(host, row, onCompact)` toggles `data-fitting` to measure the row's natural width against the container. Wide and compact states share one labelled `<nav>` element and its links (`a[aria-current="page"]` marks the current page). Put `[navbar-toggle]` on a disclosure button with `aria-controls` and `aria-expanded` rather than using an ARIA `menu`. Without JavaScript, the links remain visible and wrap.

Use `[navbar]` with `[navbar-row]`, `[navbar-brand]`, `<nav>`, `[navbar-toggle]`, and `[navbar-end]`. Optional modifiers are `[navbar-align="center"]`, `[navbar-align="end"]`, `[navbar-vivid]`, and `[navbar-sticky]`. `mountNavbar(host, {open, onOpenChange})` sets `data-compact` when the row overflows and toggles `data-open` on the `<nav>` row. `[navbar-brand]`, `[navbar-toggle]`, and `[navbar-end]` stay on the first row while `<nav>` expands below them. `mountNavbar` returns `{setOpen, dispose}` and closes the compact `<nav>` on Escape, link clicks, or outside clicks. Call `mountNavbar` on a `[navbar-sticky]` navbar even without `[navbar-toggle]`. It writes the navbar's height to `--navbar-size` on `:root` for `scroll-padding-block-start` and removes `--navbar-size` on `dispose()`.

`[responsive-toolbar]` and `[breadcrumb]` use the same `data-compact` and `data-fitting` attributes with their own compact markup and focus handling. A `[responsive-toolbar]` contains `[toolbar-actions]`, `[toolbar-group]`, `[toolbar-divider]`, `[toolbar-shortcut]`, and `[toolbar-menu]`, switching from `[toolbar-actions]` to `[toolbar-menu]` when compact. A `[breadcrumb]` contains `[breadcrumb-row]`, `[breadcrumb-parents]`, `[breadcrumb-collapse]`, `[breadcrumb-ancestor]`, and an ellipsized `[breadcrumb-current]`, with spacing controlled by `--breadcrumb-gap` (default `--space-3`). When compact, `[breadcrumb-collapse]` replaces `[breadcrumb-parents]`.

```
#| hide
#| asset styles.css
_css = r'''
@layer components {
@scope (:root, [ui]) to ([unstyle]) {
  :where(:scope):root:has([navbar][navbar-sticky]) { scroll-padding-block-start: calc(var(--navbar-size, 0px) + var(--space)); }
  :where(:scope, *)[navbar] {
    padding: var(--space-2) var(--space); border-block-end: 1px solid var(--color-border);
    background: var(--color-background); color: var(--color-text);
    &[navbar-sticky] { position: sticky; inset-block-start: 0; z-index: 10; }
    &[navbar-vivid] {
      --face: var(--color-accent); background: var(--face); color: var(--color-on-accent);
      > [navbar-row] > nav > a[aria-current=page] { background: var(--tint); color: inherit; }
    }
    &[navbar-align=center] > [navbar-row] > nav { margin-inline: auto; }
    &[navbar-align=end] > [navbar-row] > nav { margin-inline-start: auto; }
    &:is([navbar-align=center], [navbar-align=end]) > [navbar-row] > [navbar-end] { margin-inline-start: 0; }
    &[data-compact]:not([data-fitting]) {
      > [navbar-row] {
        > [navbar-toggle] { display: inline-flex; }
        > [navbar-end] { margin-inline-start: auto; }
        > nav { display: none; order: 1; flex-basis: 100%; margin: 0; }
      }
      &[data-open] > [navbar-row] > nav { display: grid; gap: var(--space-2); }
    }
    &[data-fitting] {
      > [navbar-row] {
        inline-size: max-content; max-inline-size: none; flex-wrap: nowrap;
        > nav { flex: none; flex-wrap: nowrap; }
      }
      &[data-compact] > [navbar-row] > [navbar-toggle] { display: inline-flex; position: absolute; }
    }
  }
  [navbar-row] {
    display: flex; flex-wrap: wrap; align-items: center; gap: var(--space-2) var(--space); min-inline-size: 0;
    > [navbar-brand] {
      display: inline-flex; align-items: center; gap: var(--space-3); flex: none; color: inherit;
      font-size: var(--text1); font-weight: var(--weight1); text-decoration: none; white-space: nowrap;
    }
    > nav {
      display: flex; flex-wrap: wrap; gap: var(--space-3);
      > a {
        display: inline-block; padding: var(--space-3) var(--space-2); border-radius: var(--radius-1);
        color: inherit; text-decoration: none;
        &:is(:hover, :focus-visible) { background: var(--tint); }
        &[aria-current=page] { background: var(--v-3); color: var(--v2); }
      }
    }
    > [navbar-end] { display: flex; align-items: center; gap: var(--space-2); flex: none; margin-inline-start: auto; }
    > [navbar-toggle] { display: none; flex: none; }
  }

  :where(:scope, *)[responsive-toolbar] {
    flex: 1; min-inline-size: 0;
    &[data-compact] {
      &:not([data-fitting]) > [toolbar-actions] { display: none; }
      > [toolbar-menu] { display: flex; align-items: center; gap: var(--space-3); }
    }
    &[data-fitting] {
      > [toolbar-actions] { inline-size: max-content; max-inline-size: none; flex: none; }
      > [toolbar-menu] { position: absolute; }
    }
  }
  [toolbar-actions], [toolbar-group] { display: flex; align-items: center; gap: var(--space-3); }
  [toolbar-group] + [toolbar-group] { border-inline-start: 1px solid var(--color-border); padding-inline-start: var(--space-2); margin-inline-start: var(--space-3); }
  [toolbar-menu] { display: none; }
  [toolbar-divider] { block-size: 1.25em; border-inline-start: 1px solid var(--color-border); margin-inline: var(--space-3); }
  [toolbar-shortcut] { margin-inline-start: auto; padding-inline-start: var(--space); color: var(--color-muted); font: inherit; font-size: var(--text-1); }

  :where(:scope, *)[breadcrumb] {
    min-inline-size: 0;
    &[data-compact] {
      &:not([data-fitting]) [breadcrumb-parents] { display: none; }
      [breadcrumb-collapse] { display: block; }
    }
    &[data-fitting] {
      [breadcrumb-collapse] { position: absolute; }
      > [breadcrumb-row] { inline-size: max-content; max-inline-size: none; }
    }
  }
  [breadcrumb-row] {
    display: flex; align-items: center; gap: var(--breadcrumb-gap, var(--space-3)); min-inline-size: 0;
    > :not([breadcrumb-current]) { flex: none; }
  }
  [breadcrumb-parents] {
    display: contents;
    > a { flex: none; white-space: nowrap; }
  }
  [breadcrumb-collapse] { display: none; }
  [breadcrumb-ancestor] { min-inline-size: 0; overflow-wrap: anywhere; }
  [breadcrumb-current] { min-inline-size: 0; overflow: hidden; text-overflow: ellipsis; white-space: nowrap; }
}
}
'''
```

## Dialogs, popovers, dropdowns, and tooltips

Use a native `<dialog dialog>` opened with `showModal()` for modal tasks that block the page and trap focus until completed or cancelled. Structured slots are `header` (`h2`, `p`), `[dialog-close]` in the top-end corner, body `div` elements, and `footer`. Add `[dialog-side="start"]` or `[dialog-side="end"]` to render a full-height drawer whose width defaults to `--drawer-size` (`24rem`) and expands to `100vw` below `40rem`.

Use native `popover="auto"` with `[popup]` or `[dropdown]` for non-modal panels. Native popovers appear in the browser's top layer without moving the element out of its `[ui]` theme. A `[dropdown]` holds ordinary links, buttons, and form inputs rather than an ARIA `menu`. Users move through it with Tab and dismiss it with Escape or an outside click. Inside `[dropdown]`, style entries with `[menu-item]` (supporting `aria-pressed="true"`, `[ui-tone="danger"]`, and disabled states), `[menu-label]`, `hr`, and `<details option-group>` for nested choices instead of cascading flyouts.

On browsers with CSS `position-area`, `[popup]` and `[dropdown]` flip to fit the viewport using `position-try-fallbacks`. `mountPopover(trigger, panel, {onOpenChange, side, align, sideOffset})` keeps `aria-expanded` in sync and calls `positionFloating`. On supporting browsers, `positionFloating` sets `anchor-name`, `position-anchor`, and `position-area`. On other engines, it computes `left`, `top`, and `--floating-available-height` in JavaScript. `mountPopover` hides the popover if the trigger scrolls out of the viewport and returns a cleanup function. When the panel has `[dropdown]`, it also focuses the first enabled item on open and closes on link clicks.

Wrap a trigger and its `[role="tooltip"]` element in `[tooltip]`, linking them with `aria-describedby`. CSS reveals the tooltip on `:hover` or `:has(:focus-visible)` and delays hiding by `150ms` so the pointer can cross into the tooltip. Set `--tooltip-gap` (default `6px`) to adjust the offset. A tooltip never traps focus, does not close on Escape, and requires no JavaScript.

```
#| hide
#| asset styles.css
_css = r'''
@layer components {
@scope (:root, [ui]) to ([unstyle]) {
  :where(:scope, *):is([dialog], [popup], [dropdown], [toast]) {
    --face: var(--color-surface);
    background: var(--face); color: var(--color-text);
    border: 1px solid var(--color-border); border-radius: var(--radius); box-shadow: var(--floating);
  }
  :where(:scope, *)[dialog] {
    flex-direction: column; gap: var(--space); border-radius: var(--radius1);
    inline-size: min(32rem, calc(100vw - 2 * var(--space))); max-inline-size: none; max-block-size: calc(100dvh - 2 * var(--space));
    overflow: auto; padding: var(--space1);
    > header {
      min-block-size: calc(1.25em + var(--space) + 2px); padding-inline-end: calc(1.25em + var(--space) * 2 + 2px);
      > div { min-inline-size: 0; }
      h2 { margin: 0; font-size: var(--text1); font-weight: var(--weight1); }
      p { margin: var(--space-3) 0 0; color: var(--color-muted); }
    }
    > [dialog-close] { position: absolute; inset-block-start: var(--space1); inset-inline-end: var(--space1); }
    > footer { display: flex; justify-content: flex-end; flex-wrap: wrap; gap: var(--space-2); }
    > div { min-block-size: 0; }
    &:is([dialog-side='start'], [dialog-side='end']) {
      inset: 0; inset-inline-start: auto; translate: none; block-size: 100dvh; max-block-size: none;
      inline-size: min(var(--drawer-size, 24rem), 100vw); border-radius: 0; border-block: 0; border-inline-end: 0;
      --dialog-entry: ui-drawer-enter; --slide: 100%;
    }
    &[dialog-side='start'] {
      inset-inline: 0 auto; border-inline: 0; border-inline-end: 1px solid var(--color-border); --slide: -100%;
      &:dir(rtl) { --slide: 100%; }
    }
    &[dialog-side='end']:dir(rtl) { --slide: -100%; }
    @media (width < 40rem) {
      &:is([dialog-side='start'], [dialog-side='end']) { inline-size: 100vw; border: 0; }
    }
  }
  dialog:where(:scope, *)[dialog] {
    &[open] { display: flex; animation: var(--dialog-entry, ui-dialog-enter) var(--motion) var(--ease); }
    &::backdrop { background: oklch(from var(--n2) l c h / .4); }
  }

  :where(:scope, *):is([popup], [dropdown]) {
    z-index: 60; overflow: auto;
    position: fixed; inset: auto; margin: 0; position-try-fallbacks: flip-block, flip-inline, flip-block flip-inline;
    position-try-order: most-block-size; max-block-size: calc(100% - 16px);
    &:popover-open { animation: ui-floating-enter var(--motion) var(--ease); }
    @supports not (position-area: center) { max-block-size: var(--floating-available-height, calc(100dvh - 16px)); }
  }
  :where(:scope, *)[popup] { inline-size: max-content; max-inline-size: min(24rem, calc(100vw - 2 * var(--space))); padding: var(--space); }
  :where(:scope, *)[dropdown] {
    min-inline-size: min(12rem, calc(100vw - 2 * var(--space))); padding: var(--space-3);
    &:popover-open { display: grid; }
    hr { inline-size: 100%; margin-block: var(--space-3); border: 0; border-block-start: 1px solid var(--color-border); }
    > :not([menu-item], hr, [option-group]) { padding: var(--space-2) var(--space-1); }
  }
  [menu-item] {
    display: flex; align-items: center; gap: var(--space-2); inline-size: 100%; padding: var(--space-2) var(--space-1);
    border: 0; background: none; box-shadow: none; font: inherit; text-align: start;
    border-radius: var(--radius-1); color: inherit; text-decoration: none; cursor: pointer; user-select: none;
    &:not(:disabled, [aria-disabled=true]):is(:hover, :focus-visible), &[aria-pressed=true] { background: var(--v-3); color: var(--v2); }
    &[ui-tone='danger'] { color: var(--color-accent); }
    &:is(:disabled, [aria-disabled=true]) { color: var(--n1); cursor: default; opacity: .55; }
  }
  [menu-label] { font-size: var(--text-1); font-weight: var(--weight1); color: var(--color-muted); padding-block: var(--space-3); }
  [option-group] {
    > summary {
      padding: var(--space-2) var(--space-1); cursor: pointer;
      > svg { vertical-align: middle; margin-inline-end: var(--space-2); }
    }
    > [menu-item] { padding-inline-start: var(--space1); }
  }

  :where(:scope, *)[tooltip] {
    display: inline-block; position: relative;
    > [role=tooltip] {
      z-index: 70; inline-size: max-content; max-inline-size: min(20rem, calc(100vw - 2 * var(--space))); overflow-wrap: anywhere;
      background: var(--n2); color: var(--n-3); border-radius: var(--radius-1); padding: var(--space-3) var(--space-2);
      font-size: var(--text-1); line-height: var(--leading); box-shadow: var(--floating); border: 0;
      position: fixed; inset: auto; position-area: top; position-try-fallbacks: flip-block; margin-block-end: var(--tooltip-gap, 6px);
      visibility: hidden; opacity: 0; transition: opacity var(--motion) var(--ease) 150ms, visibility 0s linear calc(var(--motion) + 150ms);
      @supports not (position-area: center) { position: absolute; inset: auto auto 100% 0; }
    }
    &:is(:hover, :has(:focus-visible)) > [role=tooltip] { visibility: visible; opacity: 1; transition: none; }
  }
  :root:has([tooltip]:is(:hover, :has(:focus-visible))) [tooltip]:not(:hover, :has(:focus-visible)) > [role=tooltip] { transition: none; }

  @keyframes ui-dialog-enter { from { opacity: 0; transform: translateY(var(--space-2)); } }
  @keyframes ui-drawer-enter { from { transform: translateX(var(--slide)); } }
  @keyframes ui-floating-enter { from { opacity: 0; transform: translateY(-2px); } }
}
}
'''
```

## Disclosures, tabs, toolbars, and toasts

Use `<details disclosure>` with `<summary>` for disclosures. Group multiple `<details disclosure>` elements inside `[accordion]` with the same `name` attribute when only one item should stay open at a time.

Mark tab structures with `[tabs]`, `[tablist]` (`role="tablist"`), `[tab]` (`role="tab"`, `data-value`, `aria-selected`, `aria-controls`), and `[tabpanel]` (`role="tabpanel"`, `hidden`). `mountTabs(root, {activationMode, onValueChange})` handles arrow keys, Home, End, and disabled tabs, returning `{setValue, dispose}`. In `'automatic'` mode (the default), moving focus to a tab with arrow keys, Home, or End selects it immediately. In `'manual'` mode, selection waits for Enter, Space, or a click. Call `setValue(value)` to update selection programmatically and `dispose()` to remove event handlers. For vertical tabs, add both `[tabs-orientation="vertical"]` on `[tabs]` and `aria-orientation="vertical"` on `[tablist]` to switch layout and arrow keys to Up and Down.

`mountToolbar(root)` sets a single roving Tab stop on buttons and links in a command row, with Left, Right, Home, and End traversal (respecting RTL). It leaves elements inside `[popover]` dropdowns in normal Tab order and returns a cleanup function. Application commands and responsive collapsing stay outside `mountToolbar`.

`mountHoverPopover(trigger, panel, {openDelay, closeDelay})` opens a popover on mouse hover after `openDelay` (default `0` ms). It closes the popover after `closeDelay` (default `300` ms) once neither element has hover or focus. Pressing `ArrowDown` on the trigger opens the panel and focuses its first enabled item. Tapping on touch also opens the panel, while the trigger's click handler stays independent (for example, a toggle button whose popover holds extra settings). `mountHoverPopover` returns a cleanup function.

Use `[toaster]` and `[toast]` for transient feedback, and place urgent messages during a modal task inside that modal. Keep the `[toaster]` live region mounted in the DOM so new messages are announced. `[toaster]` is fixed at the top-end corner by default, supports `[popover]` top-layer rendering, and accepts `[toaster-position="top-start"]`, `"bottom-start"`, or `"bottom-end"`. CSS runs the `ui-toast-lifetime` animation on each `[toast]` (with optional non-neutral `[ui-tone]`) for `--toast-duration` and pauses it on `:hover` or `:focus-within`. Remove the toast from state or the DOM when that animation ends.

```
#| hide
#| asset styles.css
_css = r'''
@layer components {
@scope (:root, [ui]) to ([unstyle]) {
  :where(:scope, *)[tabs] {
    display: grid; gap: var(--space);
    &[tabs-orientation='vertical'] {
      grid-template-columns: auto minmax(0, 1fr);
      [tablist] { flex-direction: column; border: 0; border-inline-end: 1px solid var(--color-border); }
      [tab] {
        text-align: start; border-block-end: 0; border-inline-end: 2px solid transparent;
        &[aria-selected='true'] { border-inline-end-color: var(--color-accent); }
      }
    }
  }
  [tablist] { display: flex; gap: var(--space-3); border-block-end: 1px solid var(--color-border); overflow-x: auto; }
  [tab] {
    flex-shrink: 0; border: 0; border-block-end: 2px solid transparent; padding: var(--space-2) var(--space);
    background: none; color: var(--color-muted); font: inherit; font-weight: var(--weight1); cursor: pointer;
    &[aria-selected='true'] { border-color: var(--color-accent); color: var(--v2); }
    &:hover:not(:disabled) { background: var(--v-3); color: var(--v2); }
    &:disabled { opacity: .5; cursor: default; }
  }
  [tabpanel] {
    min-inline-size: 0;
    &[hidden] { display: none; }
  }
  :where(:scope, *)[accordion] { display: grid; gap: var(--space-2); }
  :where(:scope, *)[disclosure] {
    border: 1px solid var(--color-border); border-radius: var(--radius); background: var(--color-surface);
    > summary {
      padding: var(--space-2) var(--space); cursor: pointer; font-weight: var(--weight1);
      &::marker { color: var(--color-muted); }
      &:hover { color: var(--v2); }
    }
    > :not(summary) { padding: var(--space-2) var(--space) var(--space); }
  }

  :where(:scope, *)[toast] {
    display: flex; align-items: flex-start; gap: var(--space); pointer-events: auto;
    padding: var(--space-2) var(--space);
    animation: ui-toast-lifetime var(--toast-duration) linear forwards;
    &:is(:hover, :focus-within) { animation-play-state: paused; }
    > div { display: grid; gap: var(--space-3); flex: 1; min-inline-size: 0; overflow-wrap: anywhere; }
    &[ui-tone]:not([ui-tone='neutral']) { --face: var(--v-3); color: var(--v2); border-color: var(--v-2); }
  }
  :where(:scope, *)[toaster] {
    position: fixed; z-index: 45; inset-block-start: var(--space); inset-inline-end: var(--space);
    display: grid; gap: var(--space-2); inline-size: min(22rem, calc(100vw - 2 * var(--space))); pointer-events: none;
    &[popover] {
      inset: auto; inset-block-start: var(--space); inset-inline-end: var(--space); margin: 0;
      padding: var(--space-3); border: 0; background: none; overflow: visible; color: inherit;
      &:not(:popover-open) { display: none; }
    }
    &:is([toaster-position='bottom-end'], [toaster-position='bottom-start']) { inset-block: auto var(--space); }
    &:is([toaster-position='top-start'], [toaster-position='bottom-start']) { inset-inline: var(--space) auto; }
  }

  @keyframes ui-toast-lifetime { from { opacity: 1; } to { opacity: 1; } }
}
}
'''
```
