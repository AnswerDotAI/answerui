A table describes relationships between rows and columns. HTML already has the structure: captions, column and row headers, body rows, and totals. We style those elements without replacing them with a data-grid API. The application chooses the columns, formats values, and handles filtering, sorting, and pagination.

```
import answerui.base
from fasthtml.common import Div, Caption, Thead, Tbody, Tfoot, Tr, Th, Td, Dl, Dt, Dd, Small
from answerui.registry import hypelt
from answerui.layout import Grid
from answerui.demo import example
```

```
from answerui.demo import demo
from answerui.surfaces import Badge
```

## Table

The base module already supplies cell padding, muted column headings, and row separators. Here we add a caption above the table, stronger row headings, and a rule before the totals. These remain in the base layer: typrose can keep its more specific treatment inside prose regions.

`Table` emits a native `table` with no wrapper or marker attribute. Use `scope="col"` and `scope="row"` on headers to describe their relationships. Mark numeric cells and their column headings with `numeric=True`. The cells align to the end and use tabular numerals. Text stays free to wrap.

```
Table = hypelt('Table', layer='base', doc='A native table; numeric cells align at the end, and compact reduces row spacing.', css=r"""
table > caption {
  text-align: start; caption-side: top; padding-block-end: var(--space-2);
  font-weight: var(--weight1); color: var(--n2);
}
table > tbody > tr > th { color: var(--n2); }
table > tfoot { border-block-start: 1px solid var(--n); font-weight: var(--weight1); }
table > tfoot > tr > th { color: var(--n2); }
table > :is(thead, tbody, tfoot) > tr > [numeric] {
  text-align: end; font-variant-numeric: tabular-nums; white-space: nowrap;
}
table[compact] > :is(thead, tbody, tfoot) > tr > :is(th, td) { padding-block: var(--space-3); }
""")
```

A caption names the table. Row headings identify each project, and the footer contains the total. The numbers are already formatted when they reach `Td`:

```
@example(Table)
def _table_example():
    return Table(Caption('Project storage'),
        Thead(Tr(Th('Project', scope='col'), Th('Files', scope='col', numeric=True), Th('Storage (GB)', scope='col', numeric=True))),
        Tbody(Tr(Th('Documentation', scope='row'), Td('1,240', numeric=True), Td('3.20', numeric=True)),
            Tr(Th('Research', scope='row'), Td('860', numeric=True), Td('5.40', numeric=True))),
        Tfoot(Tr(Th('Total', scope='row'), Td('2,100', numeric=True), Td('8.60', numeric=True))))
```

```
Table.demo()
```

## Narrow containers

A wide table needs a scrollable parent. Changing the table itself to a block or turning its rows into cards would obscure the column relationships. `TableScroll` adds one wrapper and leaves the table intact. It has no surface or outer spacing.

The wrapper has `tabindex="0"` so keyboard users can focus and scroll it in browsers that do not make scroll regions focusable automatically. `label` names the region. This deliberately adds a tab stop even when the table fits; there is no resize observer or overflow state to maintain. See [MDN's guidance on scroll-region accessibility](https://developer.mozilla.org/en-US/docs/Web/CSS/Reference/Properties/overflow#accessibility).

```
@hypelt(sel='[table-scroll]', layer='layout', css=r"""
[table-scroll] { min-inline-size: 0; max-inline-size: 100%; overflow-x: auto; }
""")
def TableScroll(
    table,   # The native table to keep intact
    label,   # Accessible name of the scroll region
    **attrs, # Attributes for the containing div
):
    "A named, keyboard-focusable scroll region around a table"
    return Div(table, table_scroll=True, role='region', aria_label=label, tabindex='0', **attrs)
```

`compact=True` reduces vertical padding without shrinking the type. Here the same table sits in a narrow region. Focus the region to scroll horizontally; the numeric columns stay aligned.

```
@example(TableScroll)
def _table_scroll_example():
    return TableScroll(_table_example()(compact=True), 'Project storage', style='max-inline-size: 16rem')
```

```
TableScroll.demo()
```

## Stat

A stat pairs a label with a prominent value and, optionally, a note explaining it. A description list gives us that relationship directly: the label is a `dt`, and the value and note belong to its `dd`. It has no surface or outer spacing of its own.

```
@hypelt(sel='[stat]', layer='components', css=r"""
[stat], [stat] > dd { display: grid; gap: var(--space-3); }
[stat] > dt { font-size: var(--text-1); color: var(--n1); }
[stat] > dd {
  font-size: var(--text3); font-weight: var(--weight1); line-height: var(--leading-1);
  font-variant-numeric: tabular-nums;
}
[stat] > dd > small { font-weight: var(--weight); line-height: var(--leading); color: var(--n1); }
""")
def Stat(
    label,            # What the value measures
    value,            # Display text or markup, formatted by the caller
    description=None, # Optional supporting text or inline markup
    **attrs,          # Attributes for the description list
):
    "A labelled value with an optional supporting note"
    return Dl(Dt(label), Dd(value, Small(description) if description is not None else None), stat=True, **attrs)
```

Formatting belongs to the application: counts, money, durations, and percentages all use the same component. Tabular numerals keep digits evenly spaced without requiring a monospace font.

```
@example(Stat)
def _stat_example():
    return Grid(Stat('Active projects', 12), Stat('Storage used', '8.2 GB', 'Of 20 GB available'))
```

```
Stat.demo()
```

Adding `card=True` boxes the stat itself, with no extra wrapper. The note can contain a badge when there is something worth highlighting; the component does not need a separate trend or severity API:

```
demo(Grid(
    Stat('Open requests', 18, Badge('4 need attention', warning=True), card=True),
    Stat('Response time', '2.4 h', 'Median over the last 7 days', card=True)))
```

## Customisation

Tables share the existing spacing, type, and colour tokens. To adjust one table, give it an application attribute and register CSS for that selector in the app layer. For example, `table[usage] > tfoot { background: var(--v-3); }` gives its totals a wash without changing other tables. `TableScroll` does not impose a minimum width; content determines when scrolling is needed.
