# data

> Tables and other displays of structured data

A table is a native `table`, styled by the base layer, and the browser already gives it its structure: a caption, a head, a body, and a foot. This module adds the frame, a raised surface that the table fills and that scrolls sideways when the table is wider than the page.


```
from fasthtml.common import Div, Caption, Thead, Tbody, Tfoot, Tr, Th, Td
import fasthtml.components as fc
from answerui.registry import hypelt
from answerui.demo import example
import answerui.base  # the base layer this frame extends
```

```
from fasthtml.common import *
from answerui.demo import demo
```

## Table

A table is a native `table`, and the base layer gives every table its look: a tinted head band with labels in small uppercase mono text, hairlines between rows, a semibold foot for totals, and a cell marked `num` aligned to the end with tabular digits. On its own a table hugs its content and scrolls sideways when it is wider than its container, so a page survives a phone with nothing added. Inside prose it fills the measure and wraps instead.

`Table()` puts the table in a frame: a raised surface like a card, bordered and rounded, which the table fills. The frame is the one part of the page allowed to scroll sideways, so a wide table never pushes the page wider than the screen, nor widens a card that sizes to its content, since the frame contains its inline size, and it carries the border, so the table's own hairlines above and below go.


```
_frame_css = r'''
[frame] {
  --face: var(--n-3); background: var(--face); overflow-x: auto; contain: inline-size;
  border: 1px solid var(--n-1); border-radius: var(--radius); box-shadow: var(--raised);
  > table { display: table; inline-size: 100%; border-block: 0; }
}
'''

@hypelt(sel='[frame]', layer='components', css=_frame_css)
def Table(*c, **attrs):
    "A table in a frame: a bordered surface it fills, which scrolls sideways when the table is wider than the page"
    return Div(fc.Table(*c, **attrs), frame=True)
```

Orders by region for a quarter, with a caption, numbers marked `num`, and totals in the foot:

```
@example(Table)
def _table_example():
    rows = [('North', '1,204', '38,410', '31.90'), ('South', '862', '27,955', '32.43'), ('West', '1,531', '50,112', '32.73')]
    def row(nm, *figs): return Tr(Td(nm), *[Td(f, num=True) for f in figs])
    head = Thead(Tr(Th('Region'), Th('Orders', num=True), Th('Revenue', num=True), Th('Average', num=True)))
    foot = Tfoot(Tr(Th('Total'), Td('3,597', num=True), Td('116,477', num=True), Td('32.38', num=True)))
    return Table(Caption('Revenue and average order in US$, Q3'), head, Tbody(*[row(*r) for r in rows]), foot)
```
