A plain table needs no component: the base layer styles a native `table` with its `caption`, `thead`, `tbody` and `tfoot`. This module adds `DataTable` for tables of figures. It gives the table a tinted head band and cells that never wrap, on a raised surface that scrolls sideways when the table doesn't fit.

## Tables

The base layer gives every `table` a plain look: a semibold head over a rule, hairlines between rows, a semibold foot for totals, a caption beneath, and a cell marked `num` aligned to the end with tabular digits. In a `prose` region, the table rules of [typrose](https://github.com/AnswerDotAI/typrose) take precedence, with their own spacing and smaller text.

`DataTable` puts a plain table in a `div` with the `datatable` attribute: a bordered, rounded surface like a card, which the table fills. Because the surface scrolls sideways and has `contain: inline-size`, a wide table never makes the page wider than the screen or widens a card that sizes to its content. On the surface, the table has a tinted head band with labels in small uppercase mono text. Each row stays on one line however narrow the screen, because cells never wrap.


```
_datatable_css = r'''
--face: var(--n-3); background: var(--face); overflow-x: auto; contain: inline-size;
border: 1px solid var(--n-1); border-radius: var(--radius); box-shadow: var(--raised);
> table { inline-size: 100%; }
:is(th, td) { white-space: nowrap; }
tbody tr:first-child > * { border-block-start: 1px solid var(--n-1); }
thead th { background: var(--n-2); color: var(--n1); font: var(--weight) var(--text-1) var(--font-mono); text-transform: uppercase; letter-spacing: .08em; border-block-end: 0; }
'''

@hypelt(sel='[datatable]', layer='components', css=_datatable_css)
def DataTable(*c, **attrs):
    "A table on a bordered surface that scrolls sideways when the table doesn't fit; attributes, such as an id or a variant, go on the surface"
    return Div(Table(*c), datatable=True, **attrs)
```

Orders by region for a quarter, with a caption, numbers marked `num`, and totals in the foot:

```
@example(DataTable)
def _datatable_example():
    rows = [('North', '1,204', '38,410', '31.90'), ('South', '862', '27,955', '32.43'), ('West', '1,531', '50,112', '32.73')]
    def row(nm, *figs): return Tr(Td(nm), *[Td(f, num=True) for f in figs])
    head = Thead(Tr(Th('Region'), Th('Orders', num=True), Th('Revenue', num=True), Th('Average', num=True)))
    foot = Tfoot(Tr(Th('Total'), Td('3,597', num=True), Td('116,477', num=True), Td('32.38', num=True)))
    return DataTable(Caption('Revenue and average order in US$, Q3'), head, Tbody(*[row(*r) for r in rows]), foot)
```

```
DataTable.demo()
```

Because the surface is the component's root, attributes passed to `DataTable`, including any variant an app declares on it, go on the surface rather than on the table inside:

```
t = DataTable(Tbody(Tr(Td('x'))), id='orders')
test_eq(t.attrs['id'], 'orders')
test_eq(t.children[0].tag, 'table')
```
