> Styled data tables (`table[table]`) and horizontally scrolling `DataTable` frames

Use `Table(..., table=True)` for a styled table with `caption`, `thead`, `tbody`, and `tfoot`. `DataTable` wraps that table in a bordered, raised frame (`[table-frame]`). Inside `DataTable`, cells stay on one line and the frame scrolls horizontally when the table is wider than its parent container.

## Tables

When you pass `table=True` to `Table`, `thead th` and `tfoot th` are semibold (`--weight1`), with horizontal borders under the header and between body and footer rows. Any `caption` appears below the table in muted `--text-1` text. Add `num=True` to a `Th` or `Td` to align it to the end with tabular digits (`font-variant-numeric: tabular-nums`). Inside `[prose]`, the prose stylesheet formats unmarked `table` elements with `.875em` text and prose spacing.

`DataTable` wraps `Table(*c, table=True)` in a `Div` with `table_frame=True` (`[table-frame]`). That container has a border, a `--raised` shadow, `overflow-x: auto`, and `contain: inline-size`. Because of `contain: inline-size`, a wide table scrolls inside the frame rather than widening the page or its containing card. Inside `[table-frame]`, `thead th` has a `--color-surface` background and cells stay on one line (`white-space: nowrap`). Keyword attributes passed to `DataTable` apply to the outer `Div`. For keyboard scrolling, pass `role='region'`, `tabindex=0`, and `aria_label` on `DataTable`. Focused frames show the `--ring` outline (`[table-frame]:focus-visible`).


```
@hypelt(sel='[table-frame]')
def DataTable(*c, **attrs):
    "A table in a bordered frame that scrolls when its contents are wider than its container"
    return Div(Table(*c, table=True), table_frame=True, **attrs)
```

Here is a quarterly regional summary using `Caption`, `num=True` on numeric cells, and totals in `Tfoot`:

```
@example(DataTable)
def _datatable_example():
    rows = [('North', '1,204', '38,410', '31.90'), ('South', '862', '27,955', '32.43'), ('West', '1,531', '50,112', '32.73')]
    def row(nm, *figs): return Tr(Td(nm), *[Td(f, num=True) for f in figs])
    head = Thead(Tr(Th('Region'), Th('Orders', num=True), Th('Revenue', num=True), Th('Average', num=True)))
    foot = Tfoot(Tr(Th('Total'), Td('3,597', num=True), Td('116,477', num=True), Td('32.38', num=True)))
    return DataTable(Caption('Revenue and average order in US$, Q3'), head, Tbody(*[row(*r) for r in rows]), foot,
        role='region', tabindex=0, aria_label='Orders by region')
```
